# U.S. Companion Animal Active Adoption Listing Inventory — Methodological & Regional Endpoint Guide

This reproducible Python companion notebook documents how to query, validate, and interpret the aggregated public companion animal adoption listing series maintained on **[Adopt'Animaux U.S. Shelter & Rescue Adoption Data](https://adoptanimaux.com/en/adoption/data/us)** and the **[United States Shelter & Rescue Adoption Directory](https://adoptanimaux.com/en/adoption/directory/us)** ([adoptanimaux.com/en](https://adoptanimaux.com/en/)).

> **Methodological Scope Guardrail:** All linked tables measure **active, location-ready companion animal adoption listing inventory** represented across published U.S. city hubs. They do **not** measure gross shelter intake, live-release outcomes, euthanasia counts, unique individual animal flow, or DNA-verified breed ancestry. For standardized national intake and outcome statistics, researchers should consult Shelter Animals Count alongside our [Public Adoption Inventory Methodology](https://adoptanimaux.com/en/adoption/data/methodology).
